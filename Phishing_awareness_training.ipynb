```python
from IPython.display import display, HTML
import textwrap


training_html = """
<div style="
    font-family: Arial, sans-serif;
    max-width: 900px;
    margin: auto;
    padding: 25px;
    line-height: 1.6;
    border: 1px solid #ddd;
    border-radius: 15px;
">

<h1>🎣 Phishing Awareness Training</h1>

<p>
Phishing is a cyberattack in which an attacker pretends to be a trusted
person, company, or service to trick users into revealing information,
opening malicious files, transferring money, or visiting unsafe websites.
</p>

<hr>

<h2>🔍 1. How to Recognize Phishing Emails</h2>

<ul>
<li><b>Suspicious sender:</b> The email address does not match the claimed organization.</li>
<li><b>Urgent language:</b> Messages such as "Act immediately" or "Your account will be closed."</li>
<li><b>Unexpected attachments:</b> Be careful with unknown documents, ZIP files, or executable files.</li>
<li><b>Suspicious links:</b> The displayed link may lead to a different website.</li>
<li><b>Requests for sensitive information:</b> Passwords, OTPs, banking details, or identity documents.</li>
<li><b>Unexpected financial requests:</b> Requests for gift cards, cryptocurrency, or bank transfers.</li>
<li><b>Unusual grammar or formatting:</b> Errors can be a warning sign, although well-written phishing messages also exist.</li>
</ul>

<h2>🌐 2. How to Recognize Fake Websites</h2>

<ul>
<li>Check the website address carefully.</li>
<li>Look for misspelled or misleading domain names.</li>
<li>Do not assume HTTPS automatically means a website is trustworthy.</li>
<li>Avoid logging in through links received unexpectedly by email or SMS.</li>
<li>When possible, type the organization's official website address yourself.</li>
</ul>

<h2>🧠 3. Social Engineering Tactics</h2>

<ul>
<li><b>Urgency:</b> Creating pressure so you act without checking.</li>
<li><b>Authority:</b> Pretending to be a manager, bank employee, police officer, or IT administrator.</li>
<li><b>Fear:</b> Threatening account suspension or other consequences.</li>
<li><b>Curiosity:</b> Using interesting documents, links, or messages to attract attention.</li>
<li><b>Impersonation:</b> Pretending to be someone the victim knows.</li>
<li><b>Reward:</b> Offering prizes, discounts, jobs, or financial benefits.</li>
</ul>

<h2>🛡️ 4. Best Practices</h2>

<ul>
<li>Never share passwords or OTPs through email, chat, or phone calls.</li>
<li>Verify unexpected requests using a separate trusted communication method.</li>
<li>Hover over links before clicking them.</li>
<li>Use multi-factor authentication (MFA).</li>
<li>Keep your operating system, browser, and security software updated.</li>
<li>Report suspicious emails to your organization's security/IT team.</li>
<li>If you clicked something suspicious, report it quickly instead of hiding the incident.</li>
</ul>

<h2>📧 5. Real-World Example</h2>

<div style="
    background:#f5f5f5;
    padding:18px;
    border-radius:10px;
    border-left:5px solid #555;
">

<b>Subject:</b> ⚠️ Your Bank Account Will Be Suspended Today

<p>
Dear Customer,<br><br>
We detected unusual activity on your account.
You must verify your account within 30 minutes or it will be permanently suspended.
</p>

<p>
<b>Verify your account:</b>
https://example-security-login.com
</p>

<p>
Thank you,<br>
Security Department
</p>

</div>

<h3>🚩 Warning Signs in This Example</h3>

<ol>
<li>Artificial time pressure: "within 30 minutes."</li>
<li>Fear-based language: "permanently suspended."</li>
<li>Unexpected request to verify an account.</li>
<li>The domain should be independently verified rather than trusted from the email.</li>
</ol>

<hr>

<h2>📝 6. Interactive Quiz</h2>

<p><b>Answer the following questions by entering A, B, C, or D.</b></p>

</div>
"""

display(HTML(training_html))


# ----------------------- QUIZ -------------------------------

questions = [
    {
        "q": "You receive an email saying your account will be deleted in 10 minutes unless you click a link. What should you do?",
        "options": {
            "A": "Click immediately",
            "B": "Forward it to everyone",
            "C": "Verify the message through the organization's official website/app",
            "D": "Reply with your password"
        },
        "answer": "C",
        "explanation": "Urgency is a common phishing tactic. Verify through a trusted channel."
    },
    {
        "q": "Which is a common warning sign of phishing?",
        "options": {
            "A": "An unexpected request for sensitive information",
            "B": "A message you were expecting",
            "C": "A normal internal announcement",
            "D": "A known contact sending a routine message"
        },
        "answer": "A",
        "explanation": "Unexpected requests for passwords, OTPs, banking information, etc. should be treated cautiously."
    },
    {
        "q": "What is the safest way to access your bank's website after receiving a suspicious email?",
        "options": {
            "A": "Click the email link",
            "B": "Use the link in the email signature",
            "C": "Search random links online",
            "D": "Use the bank's known official website or app"
        },
        "answer": "D",
        "explanation": "Avoid unexpected links. Use a known trusted route to access the service."
    },
    {
        "q": "An attacker pretends to be your manager and asks you to urgently purchase gift cards. What tactic is this?",
        "options": {
            "A": "Social engineering / impersonation",
            "B": "Software update",
            "C": "Data backup",
            "D": "Encryption"
        },
        "answer": "A",
        "explanation": "Attackers commonly impersonate trusted people and create urgency."
    },
    {
        "q": "What should you do if you accidentally click a suspicious phishing link?",
        "options": {
            "A": "Ignore it",
            "B": "Report it promptly to the appropriate IT/security team",
            "C": "Share the link with friends",
            "D": "Enter your password to see what happens"
        },
        "answer": "B",
        "explanation": "Prompt reporting can help protect you and others and allows the organization to respond."
    }
]

score = 0

print("\n" + "=" * 65)
print("🎯 PHISHING AWARENESS QUIZ")
print("=" * 65)

for i, item in enumerate(questions, 1):
    print(f"\nQuestion {i}/{len(questions)}")
    print(item["q"])

    for key, value in item["options"].items():
        print(f"  {key}. {value}")

    while True:
        choice = input("Your answer (A/B/C/D): ").strip().upper()
        if choice in ["A", "B", "C", "D"]:
            break
        print("Please enter A, B, C, or D.")

    if choice == item["answer"]:
        score += 1
        print("✅ Correct!")
    else:
        print(f"❌ Incorrect. Correct answer: {item['answer']}")

    print("💡", item["explanation"])


# -------------------- FINAL RESULT ---------------------------

percentage = (score / len(questions)) * 100

print("\n" + "=" * 65)
print("🏁 TRAINING COMPLETE")
print("=" * 65)
print(f"Score: {score}/{len(questions)} ({percentage:.0f}%)")

if percentage == 100:
    print("🎉 Excellent! You identified all of the phishing scenarios correctly.")
elif percentage >= 60:
    print("👍 Good work! Review the warning signs and social-engineering tactics once more.")
else:
    print("📚 Review the training material and try the quiz again.")

print("\nRemember:")
print("• Think before clicking.")
print("• Verify unexpected requests.")
print("• Never share passwords or OTPs.")
print("• Report suspicious messages.")
print("=" * 65)
```
