```python
from scapy.all import sniff, IP, IPv6, TCP, UDP, ICMP, Raw, ARP
from datetime import datetime

PACKET_COUNT = 20
INTERFACE = None

def analyze_packet(packet):
    timestamp = datetime.now().strftime("%H:%M:%S")

    # ARP packet
    if packet.haslayer(ARP):
        print(f"\n[{timestamp}] ARP")
        print(f"  Source MAC      : {packet[ARP].hwsrc}")
        print(f"  Destination MAC : {packet[ARP].hwdst}")
        print(f"  Source IP       : {packet[ARP].psrc}")
        print(f"  Destination IP  : {packet[ARP].pdst}")
        print(f"  Operation       : {packet[ARP].op}")
        return

    # IPv4 packet
    if packet.haslayer(IP):
        src_ip = packet[IP].src
        dst_ip = packet[IP].dst
        protocol_number = packet[IP].proto

    # IPv6 packet
    elif packet.haslayer(IPv6):
        src_ip = packet[IPv6].src
        dst_ip = packet[IPv6].dst
        protocol_number = packet[IPv6].nh

    else:
        return

    # Identify transport protocol
    if packet.haslayer(TCP):
        protocol = "TCP"
        src_port = packet[TCP].sport
        dst_port = packet[TCP].dport

    elif packet.haslayer(UDP):
        protocol = "UDP"
        src_port = packet[UDP].sport
        dst_port = packet[UDP].dport

    elif packet.haslayer(ICMP):
        protocol = "ICMP"
        src_port = "-"
        dst_port = "-"

    else:
        protocol = f"IP ({protocol_number})"
        src_port = "-"
        dst_port = "-"

    print(f"\n[{timestamp}] Packet")
    print(f"  Source IP       : {src_ip}")
    print(f"  Destination IP  : {dst_ip}")
    print(f"  Protocol        : {protocol}")
    print(f"  Source Port     : {src_port}")
    print(f"  Destination Port: {dst_port}")
    print(f"  Packet Size     : {len(packet)} bytes")

    # Display payload when present
    if packet.haslayer(Raw):
        payload = bytes(packet[Raw].load)

        # Safely display printable characters
        preview = ''.join(
            chr(b) if 32 <= b <= 126 else '.'
            for b in payload[:100]
        )

        print(f"  Payload Length  : {len(payload)} bytes")
        print(f"  Payload Preview : {preview}")
    else:
        print("  Payload         : None")

print("=" * 70)
print("NETWORK PACKET CAPTURE & ANALYSIS")
print("=" * 70)
print(f"Capturing up to {PACKET_COUNT} packets...")
print("Press Ctrl+C to stop early.\n")

try:
    sniff(
        iface=INTERFACE,
        prn=analyze_packet,
        count=PACKET_COUNT,
        store=False
    )
except PermissionError:
    print("\nPermission denied. Run the program as Administrator/root.")
except KeyboardInterrupt:
    print("\nCapture stopped by user.")
except Exception as e:
    print(f"\nCapture error: {e}")

print("\n" + "=" * 70)
print("CAPTURE COMPLETE")
print("=" * 70)
```
